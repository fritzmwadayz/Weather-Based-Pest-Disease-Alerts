```python
import pandas as pd
import numpy as np
```

```python
# Random seed for reproducibility
np.random.seed(42)

# Number of samples
n_samples = 5000  # Data points

# Weather data for rice
temperature = np.random.uniform(20, 35, n_samples)  # Temperature between 20°C and 35°C
humidity = np.random.uniform(60, 95, n_samples)     # Humidity between 60% and 95%import pandas as pd
import numpy as np
rainfall = np.random.uniform(0, 50, n_samples)      # Rainfall between 0mm and 50mm
wind_speed = np.random.uniform(0, 20, n_samples)    # Wind speed between 0km/h and 20km/h
```

```python
# Simulate pest/disease presence for rice
# Blast: Thrives in warm, humid conditions
blast = ((temperature > 25) & (humidity > 80)).astype(int)

# Brown Spot: Thrives in warm, wet conditions
brown_spot = ((temperature > 25) & (rainfall > 30)).astype(int)

# Stem Borers: Thrive in warm, humid conditions
stem_borers = ((temperature > 28) & (humidity > 70)).astype(int)
```

```python
# Combine into a DataFrame
rice_data = pd.DataFrame({
    'Temperature': temperature,
    'Humidity': humidity,
    'Rainfall': rainfall,
    'Wind_Speed': wind_speed,
    'Blast': blast,
    'Brown_Spot': brown_spot,
    'Stem_Borers': stem_borers
})

# Save the dataset to a CSV file
rice_data.to_csv('data/rice_pest_data.csv', index=False)
```

```python
#Model:
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import pickle
```

```python
# Features and target variables for rice
X_rice = rice_data[['Temperature', 'Humidity', 'Rainfall', 'Wind_Speed']]
y_rice = rice_data[['Blast', 'Brown_Spot', 'Stem_Borers']]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_rice, y_rice, test_size=0.2, random_state=42)

# Train a multi-label model
model_rice = MultiOutputClassifier(RandomForestClassifier(random_state=42))
model_rice.fit(X_train, y_train)
```

```python
# Model evaluation:
y_pred = model_rice.predict(X_test)
print("Rice Model Evaluation:")
for i, pest in enumerate(['Blast', 'Brown_Spot', 'Stem_Borers']):
    print(f"{pest}:")
    print(classification_report(y_test.iloc[:, i], y_pred[:, i]))
```

```python
# Save the model using pickle
with open('rice_pest_model.pkl', 'wb') as f:
    pickle.dump(model_rice, f)
```
